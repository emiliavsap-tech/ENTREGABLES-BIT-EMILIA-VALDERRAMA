2. LIMPIEZA DE DATOS

En esta etapa se realizarán las transformaciones necesarias para mejorar la calidad y consistencia del conjunto de datos. Se tratarán los valores faltantes, se eliminarán los registros duplicados, se corregirán los tipos de datos y se estandarizarán las variables categóricas.

El propósito es obtener un DataFrame limpio y confiable para realizar los análisis posteriores.

2.1 Identificación y tratamiento de valores faltantes

Se crea una copia del DataFrame original para conservar los datos sin modificar. Posteriormente, se evaluará la cantidad de valores faltantes y se definirá una estrategia de tratamiento según la importancia de cada variable y la cantidad de información disponible.

```python
# Crear una copia del dataset original
df_limpio = df.copy()

# Revisar los valores faltantes antes de la limpieza
print("VALORES FALTANTES ANTES DE LA LIMPIEZA")
print(df_limpio.isnull().sum())

# Revisar el porcentaje de datos faltantes
print("\nPORCENTAJE DE VALORES FALTANTES")
print((df_limpio.isnull().mean() * 100).round(2))

# Revisar las estadísticas de las variables numéricas
print("\nESTADÍSTICAS DE LAS VARIABLES NUMÉRICAS")
display(df_limpio.describe())
```

2.2 Estrategia para el tratamiento de valores faltantes

De acuerdo con el diagnóstico realizado, se establecen las siguientes estrategias:

Se eliminará la columna Animal code, debido a que no contiene información en ninguno de sus registros.

Se conservará Animal name y los valores faltantes se identificarán como "Desconocido", para preservar los registros disponibles.

En las variables Animal type y Country, los valores ausentes se reemplazarán por "Desconocido".

En Gender, los valores faltantes se reemplazarán por "not determined".

Las variables numéricas Weight kg y Body Length cm se revisarán antes de imputar sus valores faltantes, debido a la presencia de valores sospechosos.

Los valores faltantes de Latitude y Longitude se conservarán, ya que no es adecuado inventar ubicaciones geográficas.

```python
# Revisar los valores negativos y extremos
print("PESOS NEGATIVOS")
display(df_limpio[df_limpio["Weight kg"] < 0])

print("\nLONGITUDES NEGATIVAS")
display(df_limpio[df_limpio["Body Length cm"] < 0])

print("\nPESOS MAYORES A 1000 KG")
display(df_limpio[df_limpio["Weight kg"] > 1000])

print("\nLONGITUDES MAYORES A 300 CM")
display(df_limpio[df_limpio["Body Length cm"] > 300])
```

2.3 Identificación y tratamiento de valores numéricos inválidos

Durante la revisión de las variables numéricas se encontraron valores negativos en las columnas Weight kg y Body Length cm. Estos valores no son físicamente posibles, por lo que se consideran registros inválidos.

Se reemplazarán por valores nulos (NaN), para posteriormente aplicar una estrategia de imputación basada en la mediana de cada tipo de animal.

También se identificaron valores extremos superiores a 1.000 kg y 300 cm. Estos se conservarán temporalmente, ya que requieren una validación adicional y no deben eliminarse sin justificación.

```python
# Reemplazar valores negativos por valores faltantes

df_limpio.loc[df_limpio["Weight kg"] < 0, "Weight kg"] = np.nan

df_limpio.loc[df_limpio["Body Length cm"] < 0, "Body Length cm"] = np.nan

# Comprobar los cambios
print("VALORES NEGATIVOS DESPUÉS DE LA CORRECCIÓN")

print("Pesos negativos:",
      (df_limpio["Weight kg"] < 0).sum())

print("Longitudes negativas:",
      (df_limpio["Body Length cm"] < 0).sum())

print("\nNUEVOS VALORES FALTANTES")
print(df_limpio[["Weight kg", "Body Length cm"]].isnull().sum())
```

2.4 Identificación y eliminación de registros duplicados

Los registros duplicados pueden generar resultados incorrectos en los análisis, ya que una misma observación podría contabilizarse más de una vez.

Por esta razón, se identifican y eliminan únicamente las filas que presentan información completamente repetida, conservando una sola copia de cada registro.

```python
# Identificar registros duplicados
duplicados = df_limpio.duplicated().sum()

print("REGISTROS DUPLICADOS ANTES DE LA LIMPIEZA:", duplicados)

# Eliminar registros duplicados
df_limpio = df_limpio.drop_duplicates()

# Comprobar el resultado
print("\nREGISTROS DESPUÉS DE ELIMINAR DUPLICADOS:")
print("Cantidad de filas:", df_limpio.shape[0])
print("Duplicados restantes:", df_limpio.duplicated().sum())
```

2.5 Estandarización de variables categóricas

Se identificaron inconsistencias en las columnas Animal type y Country. Algunos animales presentan errores de escritura y algunos países aparecen representados mediante abreviaturas.

Para mejorar la calidad de los datos, se unificarán las categorías equivalentes, conservando una representación estándar para cada valor.

```python
df_limpio["Animal type"] = df_limpio["Animal type"].str.strip()

df_limpio["Animal type"] = df_limpio["Animal type"].replace({
    "European buster": "European bison",
    "European bisson": "European bison",
    "European bison™": "European bison",
    "ledgehog": "hedgehog",
    "wedgehod": "hedgehog"
})

# Estandarizar nombres de países

df_limpio["Country"] = df_limpio["Country"].str.strip()

df_limpio["Country"] = df_limpio["Country"].replace({
    "PL": "Poland",
    "HU": "Hungary",
    "CZ": "Czech Republic",
    "DE": "Germany"
})

# Comprobar las categorías corregidas

print("TIPOS DE ANIMALES:")
print(df_limpio["Animal type"].value_counts(dropna=False))

print("\nPAÍSES:")
print(df_limpio["Country"].value_counts(dropna=False))
```

```python
# Corregir las categorías que aún presentan inconsistencias

df_limpio["Animal type"] = df_limpio["Animal type"].replace({
    "red squirrell": "red squirrel",
    "lynx?": "lynx",
    "ledgehod": "hedgehog"
})

df_limpio["Country"] = df_limpio["Country"].replace({
    "Hungry": "Hungary",
    "Czech": "Czech Republic"
})

# Verificar nuevamente

print("ANIMALES CORREGIDOS:")
print(df_limpio["Animal type"].value_counts(dropna=False))

print("\nPAÍSES CORREGIDOS:")
print(df_limpio["Country"].value_counts(dropna=False))
```

```python
print("NOMBRES EXACTOS DE LOS ANIMALES:")
for valor in df_limpio["Animal type"].dropna().unique():
    print(repr(valor))

print("\nNOMBRES EXACTOS DE LOS PAÍSES:")
for valor in df_limpio["Country"].dropna().unique():
    print(repr(valor))
```

```python
# Unificar todas las variantes del nombre de la ardilla

mascara = df_limpio["Animal type"].str.contains(
    "red", case=False, na=False
) & df_limpio["Animal type"].str.contains(
    "squir", case=False, na=False
)

df_limpio.loc[mascara, "Animal type"] = "red squirrel"

# Verificar resultado
print(df_limpio["Animal type"].value_counts(dropna=False))
```

2.6 Verificación de valores faltantes y tipos de datos después de la limpieza

En esta etapa se revisa nuevamente la cantidad de valores faltantes y los tipos de datos de cada columna del DataFrame. Esta verificación permite identificar qué problemas persisten y determinar el tratamiento adecuado para completar el proceso de limpieza.

```python
# Verificar valores faltantes después de las correcciones

print("VALORES FALTANTES ACTUALES")
print(df_limpio.isnull().sum())

print("\nTIPOS DE DATOS ACTUALES")
print(df_limpio.dtypes)
```

2.7 Eliminación de columnas con alta proporción de valores faltantes

Se identificaron las columnas Animal code y Animal name con una proporción muy elevada de datos ausentes. Debido a que esta información no ofrece suficiente cobertura para realizar análisis confiables, se decide excluir ambas columnas del conjunto de datos.

```python
# Eliminar columnas con demasiados valores faltantes

df_limpio = df_limpio.drop(
    columns=["Animal code", "Animal name"]
)

# Verificar las columnas eliminadas

print("COLUMNAS DESPUÉS DE LA ELIMINACIÓN")
print(df_limpio.columns.tolist())

print("\nDIMENSIONES ACTUALES")
print(df_limpio.shape)
```

2.8 Tratamiento de valores faltantes en variables categóricas

Se identificaron valores faltantes en las columnas Animal type, Country y Gender. Para conservar los registros y evitar eliminar observaciones útiles, se reemplazarán los valores ausentes por la categoría "Unknown" (desconocido).

```python
# Reemplazar valores faltantes en variables categóricas

columnas_categoricas = ["Animal type", "Country", "Gender"]

for columna in columnas_categoricas:
    df_limpio[columna] = df_limpio[columna].fillna("Unknown")

# Verificar los cambios

print("VALORES FALTANTES EN VARIABLES CATEGÓRICAS")
print(df_limpio[columnas_categoricas].isnull().sum())

print("\nCANTIDAD DE REGISTROS DESCONOCIDOS")
for columna in columnas_categoricas:
    print(columna, ":", (df_limpio[columna] == "Unknown").sum())
```

2.9 Tratamiento de valores faltantes en variables numéricas

Se utilizará la mediana para reemplazar los valores faltantes de las columnas Weight kg y Body Length cm. Esta estrategia permite conservar los registros y reducir la influencia de valores extremos.

```python
# Imputar valores faltantes con la mediana

columnas_numericas = ["Weight kg", "Body Length cm"]

for columna in columnas_numericas:
    mediana = df_limpio[columna].median()
    df_limpio[columna] = df_limpio[columna].fillna(mediana)

# Verificar los cambios

print("VALORES FALTANTES DESPUÉS DE LA IMPUTACIÓN")
print(df_limpio[columnas_numericas].isnull().sum())

print("\nMEDIANAS UTILIZADAS")
print(df_limpio[columnas_numericas].median())
```

2.10 Conversión y verificación de tipos de datos

Se revisarán los tipos de datos de las columnas del DataFrame para asegurar que las variables numéricas tengan formato numérico y las fechas puedan utilizarse correctamente en futuros análisis.

```python
# Convertir la columna de fecha al formato datetime

df_limpio["Observation date"] = pd.to_datetime(
    df_limpio["Observation date"],
    errors="coerce",
    dayfirst=True
)

# Verificar los tipos de datos

print("TIPOS DE DATOS DESPUÉS DE LA CONVERSIÓN")
print(df_limpio.dtypes)

print("\nVALORES FALTANTES EN FECHAS")
print(df_limpio["Observation date"].isnull().sum())
```

2.11 Identificación de fechas con formato inconsistente
Se identificaron 35 fechas que no pudieron convertirse al formato datetime. Se revisarán los valores originales para identificar formatos inconsistentes o fechas inválidas antes de definir su tratamiento.

```python
# Identificar las fechas que no pudieron convertirse

fechas_invalidas = df_limpio[
    df_limpio["Observation date"].isnull()
].index

print("FECHAS ORIGINALES QUE NO SE PUDIERON CONVERTIR")

print(df.loc[fechas_invalidas, "Observation date"].value_counts())
```

2.12 Corrección de formatos inconsistentes en las fechas

Se encontraron fechas con formatos diferentes, incluyendo meses escritos en inglés y fechas separadas por puntos. Se aplicará una conversión flexible para interpretar los formatos y se revisarán nuevamente los valores faltantes.

```python
# Corregir formatos inconsistentes de fechas

fechas_originales = df["Observation date"]

# Primera conversión: interpretar fechas considerando día primero
fechas_corregidas = pd.to_datetime(
    fechas_originales,
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

# Segunda conversión: intentar interpretar las fechas que quedaron pendientes
fechas_alternativas = pd.to_datetime(
    fechas_originales,
    format="mixed",
    dayfirst=False,
    errors="coerce"
)

# Recuperar las fechas que no se convirtieron en el primer intento
fechas_corregidas = fechas_corregidas.fillna(fechas_alternativas)

# Actualizar la columna
df_limpio["Observation date"] = fechas_corregidas

# Verificar resultados
print("FECHAS FALTANTES DESPUÉS DE LA CORRECCIÓN")
print(df_limpio["Observation date"].isnull().sum())

print("\nTIPO DE DATO")
print(df_limpio["Observation date"].dtype)
```

Se realiza una verificación general del DataFrame después de aplicar las transformaciones. Se revisan sus dimensiones, valores faltantes, registros duplicados y tipos de datos para comprobar la calidad de la información

```python
# Verificación final del DataFrame limpio

print("DIMENSIONES DEL DATAFRAME")
print("Filas:", df_limpio.shape[0])
print("Columnas:", df_limpio.shape[1])

print("\nVALORES FALTANTES POR COLUMNA")
print(df_limpio.isnull().sum())

print("\nREGISTROS DUPLICADOS")
print(df_limpio.duplicated().sum())

print("\nTIPOS DE DATOS")
print(df_limpio.dtypes)

print("\nPRIMERAS 5 FILAS DEL DATAFRAME LIMPIO")
display(df_limpio.head())
```

3. NumPy

3. NumPy: Transformaciones y cálculos estadísticos

Se utilizará la biblioteca NumPy para realizar transformaciones y cálculos estadísticos sobre las variables numéricas del conjunto de datos limpio. Se creará una nueva variable mediante una condición y se calculará una medida estadística para analizar el peso de los animales.

3.1 Clasificación de animales según su peso

Se clasificará a los animales en dos grupos según su peso: animales de peso superior a la mediana y animales de peso igual o inferior a la mediana. Para ello se utilizará la función np.where().

```python
# Clasificar animales según su peso utilizando np.where()

mediana_peso = np.nanmedian(df_limpio["Weight kg"])

df_limpio["Clasificacion peso"] = np.where(
    df_limpio["Weight kg"] > mediana_peso,
    "Peso superior a la mediana",
    "Peso igual o inferior a la mediana"
)

# Verificar la nueva variable

print("CLASIFICACIÓN DE LOS ANIMALES SEGÚN SU PESO")
print(df_limpio["Clasificacion peso"].value_counts())
```

3.2 Cálculo de medidas estadísticas del peso de los animales

Se utilizarán las funciones estadísticas de NumPy para calcular el peso promedio y la desviación estándar de los animales. Estas medidas permiten conocer el peso medio y la dispersión de los datos respecto al promedio.

```python
# Calcular medidas estadísticas utilizando NumPy

peso_promedio = np.nanmean(df_limpio["Weight kg"])
desviacion_peso = np.nanstd(df_limpio["Weight kg"])

# Mostrar resultados

print("ANÁLISIS ESTADÍSTICO DEL PESO")
print("Peso promedio:", round(peso_promedio, 2), "kg")
print("Desviación estándar:", round(desviacion_peso, 2), "kg")
```

Interpretación: El peso promedio de los animales registrados es de 46,29 kg. La desviación estándar es elevada, lo que indica que existen diferencias importantes entre los pesos. Esto puede deberse a valores extremos dentro del conjunto de datos.

4. Análisis de datos con groupby()

4. Análisis de datos agrupados con groupby()

Una vez realizada la limpieza de los datos, se utilizará la función groupby() de Pandas para agrupar los registros y obtener información sobre los tipos de animales, sus pesos y los países donde fueron observados. Esto permitirá identificar patrones y comparar las características de los animales registrados.

4.1 Número de observaciones por tipo de animal

Se calculará la cantidad de observaciones registradas para cada tipo de animal, con el propósito de identificar cuáles tienen mayor presencia en el conjunto de datos

```python
# Contar observaciones por tipo de animal

observaciones_animales = df_limpio.groupby(
    "Animal type"
).size().sort_values(ascending=False)

print("NÚMERO DE OBSERVACIONES POR TIPO DE ANIMAL")
print(observaciones_animales)
```

4.2, 4.3 y 4.4. Análisis comparativo de las características de los animales por especie y país

Se analizarán los datos mediante groupby() para calcular el peso promedio por tipo de animal, el peso promedio por país y la longitud corporal promedio por especie. Estos resultados permitirán comparar las características físicas de los animales y observar diferencias entre los países.

```python
# 4.2 Peso promedio según el tipo de animal

peso_por_animal = df_limpio.groupby(
    "Animal type"
)["Weight kg"].mean().sort_values(ascending=False)

print("PESO PROMEDIO POR TIPO DE ANIMAL")
print(peso_por_animal.round(2))


# 4.3 Peso promedio según el país

peso_por_pais = df_limpio.groupby(
    "Country"
)["Weight kg"].mean().sort_values(ascending=False)

print("\nPESO PROMEDIO POR PAÍS")
print(peso_por_pais.round(2))


# 4.4 Longitud corporal promedio según el tipo de animal

longitud_por_animal = df_limpio.groupby(
    "Animal type"
)["Body Length cm"].mean().sort_values(ascending=False)

print("\nLONGITUD CORPORAL PROMEDIO POR TIPO DE ANIMAL")
print(longitud_por_animal.round(2))
```
